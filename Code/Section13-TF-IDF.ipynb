```python
import jieba  
from sklearn.feature_extraction.text import TfidfVectorizer  
import numpy as np  
import pandas as pd  

# 示例新闻短文本  
documents = [  
 'This is the first document.',

'This document is the second document.',

'And this is the third one.',

'Is this the first document?',
]  

# 使用 jieba 进行分词  
def cut_words(text):  
    return " ".join(jieba.cut(text))  

# 分词处理  
cut_documents = [cut_words(doc) for doc in documents]  
print("分词结果:")  
for i, doc in enumerate(cut_documents):  
    print(f"文档 {i+1}: {doc}")  

# 创建TF-IDF向量化器  
vectorizer = TfidfVectorizer()  

# 转换分词后的文档为TF-IDF矩阵  
tfidf_matrix = vectorizer.fit_transform(cut_documents)  

# 获取特征名称和对应的TF-IDF值  
feature_names = vectorizer.get_feature_names()  
tfidf_values = tfidf_matrix.toarray()  

# 计算每个词的TF-IDF值，并将其与词汇结合  
tfidf_results = {}  
for i in range(len(documents)):  
    for j in range(len(feature_names)):  
        #if tfidf_values[i, j] >= 0:  
        tfidf_results[feature_names[j]] = tfidf_results.get(feature_names[j], 0) + tfidf_values[i, j]  

# 将结果排序  
sorted_tfidf = sorted(tfidf_results.items(), key=lambda item: item[1], reverse=True)  

# 输出排序结果  
print("\n所有文档中每个词的TF-IDF值排序:\n")  
for word, score in sorted_tfidf:  
    print(f"{word}: {score:.4f}")
```
